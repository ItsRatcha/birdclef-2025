```python
import os
import librosa
import librosa.display
import matplotlib.pyplot as plt
import soundfile as sf
import numpy as np
```

```python
target_id = '41970'  # Change this to any folder name in train_audio/
use_mfcc = True         # Set to True for MFCC instead of Mel spectrogram
max_files = 10           # Limit how many to process
```

```python
input_dir = os.path.join('train_audio', target_id)
output_dir = os.path.join('spectrograms', target_id)
os.makedirs(output_dir, exist_ok=True)
```

```python
# === PROCESS AUDIO FILES ===
for i, filename in enumerate(os.listdir(input_dir)):
    if not filename.endswith('.ogg') or i >= max_files:
        continue

    filepath = os.path.join(input_dir, filename)
    y, sr = sf.read(filepath)
    if len(y.shape) > 1:
        y = y[:, 0]  # handle stereo

    y = librosa.util.fix_length(y, size=sr*5)  # pad/trim to 5 sec

    plt.figure(figsize=(10, 4))
    if use_mfcc:
        mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
        librosa.display.specshow(mfcc, x_axis='time')
        plt.title('MFCC')
    else:
        mel = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128)
        mel_db = librosa.power_to_db(mel, ref=np.max)
        librosa.display.specshow(mel_db, sr=sr, x_axis='time', y_axis='mel')
        plt.title('Mel Spectrogram')

    plt.colorbar(format='%+2.0f dB')
    plt.tight_layout()
    output_path = os.path.join(output_dir, f'{os.path.splitext(filename)[0]}.png')
    plt.savefig(output_path)
    plt.close()

print(f"Saved to {output_dir}")
```
